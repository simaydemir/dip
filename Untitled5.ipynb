```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Kendi Gri Tonlama (Grayscale) Fonksiyonumuz
def my_bgr2gray(img_bgr):
    # BGR formatındaki resmi float tipine dönüştür
    b = img_bgr[:, :, 0].astype(np.float32)
    g = img_bgr[:, :, 1].astype(np.float32)
    r = img_bgr[:, :, 2].astype(np.float32)

    # Standart piksel ağırlık formülü: Y = 0.299*R + 0.587*G + 0.114*B
    gray_custom = 0.299 * r + 0.587 * g + 0.114 * b

    # Değerleri uint8 (0-255 piksel) formatına yuvarla
    return np.round(gray_custom).astype(np.uint8)

# 2. Resim Okuma
img = cv2.imread('resim.jpg')

if img is None:
    print("HATA: 'resim.jpg' bulunamadı! Sol taraftaki klasöre yüklediğinden emin ol.")
else:
    # A) OpenCV hazır fonksiyonu ile gri tonlama
    gray_opencv = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # B) Kendi fonksiyonumuz ile gri tonlama
    gray_custom = my_bgr2gray(img)

    # 3. Mutlak Farkların Toplamı (Hata Değeri Hesaplama)
    # Pikseller arasındaki mutlak fark alınır ve toplanır
    absolute_difference = np.abs(gray_opencv.astype(np.int32) - gray_custom.astype(np.int32))
    error_value = np.sum(absolute_difference)

    print(f"--- RAPOR ---")
    print(f"Toplam Mutlak Fark (Hata Değeri): {error_value}")
    print(f"Piksel Başına Ortalama Hata: {np.mean(absolute_difference):.4f}")

    # 4. Görselleri Çizdirme
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(gray_opencv, cmap='gray')
    plt.title('OpenCV cvtColor Çıktısı')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(gray_custom, cmap='gray')
    plt.title('Kendi Fonksiyonumuzun Çıktısı')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
