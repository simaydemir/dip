```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# --- Temel Sinyal Üretim Fonksiyonları ---
def generate_sinusoidal_signal_gpu(size_power_of_two, amplitude, frequency, phase=0.0, sampling_rate=1000):
    N = int(2**size_power_of_two)
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    t_gpu = torch.linspace(0, N / sampling_rate, N, device=device, dtype=torch.float32)
    angles_gpu = 2 * np.pi * frequency * t_gpu + phase
    sin_wave_gpu = amplitude * torch.sin(angles_gpu)
    return t_gpu, sin_wave_gpu, sampling_rate

def add_noise_to_vector_gpu(signal_tensor, noise_ratio):
    noise_ratio = max(0.0, min(1.0, noise_ratio))
    max_amplitude = torch.max(torch.abs(signal_tensor)).item()
    std_dev = max_amplitude * noise_ratio
    noise = torch.randn_like(signal_tensor) * std_dev
    return signal_tensor + noise

# --- 6. Görev Fonksiyonu: Frekans Filtresi (Thresholding + IDFT) ---
def frequency_threshold_filter_gpu(noisy_signal_tensor, threshold_ratio=0.2):
    """
    Frekans uzayında eşik değerin altında kalan harmonikleri sıfırlayarak
    Ters DFT (IDFT) ile temizlenmiş sinyali yeniden üreten fonksiyon.
    """
    # 1. GPU üzerinde FFT (DFT) dönüşümü
    fft_complex = torch.fft.fft(noisy_signal_tensor)

    # 2. Genlik (Harmonik) hesabı ve eşik değer belirleme
    magnitudes = torch.abs(fft_complex)
    max_mag = torch.max(magnitudes)
    threshold_value = max_mag * threshold_ratio

    # 3. Eşik değerin altında kalan harmonikleri filtreleme (0 yapma)
    fft_filtered = fft_complex.clone()
    fft_filtered[magnitudes < threshold_value] = 0

    # 4. Ters FFT (IDFT) ile zaman ortamına geri dönüşüm
    reconstructed_signal = torch.fft.ifft(fft_filtered).real

    return reconstructed_signal, fft_complex, fft_filtered

# --- 5. Görevdeki Sinyallerin Hazırlanması ---
power_of_two = 10
sampling_rate = 1000
noise_level = 0.35  # %35 gürültü

# Sinyal 1 (50 Hz, Genlik 12) ve Sinyal 2 (30 Hz, Genlik 5)
t_gpu, s1_clean, _ = generate_sinusoidal_signal_gpu(power_of_two, 12.0, 50.0, sampling_rate=sampling_rate)
_, s2_clean, _ = generate_sinusoidal_signal_gpu(power_of_two, 5.0, 30.0, sampling_rate=sampling_rate)

s1_noisy = add_noise_to_vector_gpu(s1_clean, noise_level)
s2_noisy = add_noise_to_vector_gpu(s2_clean, noise_level)

# Toplam Sinyal
s_total_clean = s1_clean + s2_clean
s_total_noisy = s1_noisy + s2_noisy

# --- Frekans Filtresinin Uygulanması (Eşik Oranı: %20) ---
s1_filtered, _, _ = frequency_threshold_filter_gpu(s1_noisy, threshold_ratio=0.20)
s2_filtered, _, _ = frequency_threshold_filter_gpu(s2_noisy, threshold_ratio=0.20)
s_total_filtered, _, _ = frequency_threshold_filter_gpu(s_total_noisy, threshold_ratio=0.20)

# CPU'ya Aktarım
t = t_gpu.cpu().numpy()
s_total_clean_np = s_total_clean.cpu().numpy()
s_total_noisy_np = s_total_noisy.cpu().numpy()
s_total_filtered_np = s_total_filtered.cpu().numpy()

# --- Görselleştirme ---
fig, axes = plt.subplots(3, 1, figsize=(12, 9), sharex=True)

# Sinyal 1
axes[0].plot(t[:200], s1_clean.cpu().numpy()[:200], 'b--', label='Orijinal Temiz (50 Hz)', alpha=0.7)
axes[0].plot(t[:200], s1_noisy.cpu().numpy()[:200], 'r', label='Gürültülü', alpha=0.4)
axes[0].plot(t[:200], s1_filtered.cpu().numpy()[:200], 'g', label='Frekans Filtresi İle Yeniden Üretilen', linewidth=2)
axes[0].set_title("1. Sinyal (f = 50 Hz, Genlik = 12) İçin Frekans Filtreleme")
axes[0].grid(True)
axes[0].legend()

# Sinyal 2
axes[1].plot(t[:200], s2_clean.cpu().numpy()[:200], 'b--', label='Orijinal Temiz (30 Hz)', alpha=0.7)
axes[1].plot(t[:200], s2_noisy.cpu().numpy()[:200], 'r', label='Gürültülü', alpha=0.4)
axes[1].plot(t[:200], s2_filtered.cpu().numpy()[:200], 'purple', label='Frekans Filtresi İle Yeniden Üretilen', linewidth=2)
axes[1].set_title("2. Sinyal (f = 30 Hz, Genlik = 5) İçin Frekans Filtreleme")
axes[1].grid(True)
axes[1].legend()

# Toplam Sinyal
axes[2].plot(t[:200], s_total_clean_np[:200], 'b--', label='Orijinal Temiz Toplam Sinyal', alpha=0.7)
axes[2].plot(t[:200], s_total_noisy_np[:200], 'r', label='Gürültülü Toplam Sinyal', alpha=0.4)
axes[2].plot(t[:200], s_total_filtered_np[:200], 'm', label='Frekans Filtresi İle Yeniden Üretilen Toplam Sinyal', linewidth=2)
axes[2].set_title("Toplam Sinyal İçin Frekans Filtreleme")
axes[2].set_xlabel("Zaman (saniye)")
axes[2].grid(True)
axes[2].legend()

plt.tight_layout()
plt.show()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# --- 1. CUDA (GPU) Cihaz Ayarı ---
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# --- 2. Yardımcı Fonksiyonlar ---
def generate_signal(N, amplitude, frequency, sampling_rate=1000):
    t = torch.linspace(0, N / sampling_rate, N, device=device, dtype=torch.float32)
    signal = amplitude * torch.sin(2 * np.pi * frequency * t)
    return t, signal

def add_noise(signal_tensor, noise_ratio=0.35):
    max_amp = torch.max(torch.abs(signal_tensor)).item()
    noise = torch.randn_like(signal_tensor) * (max_amp * noise_ratio)
    return signal_tensor + noise

# --- 3. Görev 6: Frekans Filtresi Fonksiyonu (Thresholding + IDFT) ---
def frequency_threshold_filter(noisy_signal, threshold_ratio=0.20):
    # GPU üzerinde FFT (DFT)
    fft_complex = torch.fft.fft(noisy_signal)

    # Eşik değer altındaki harmonikleri sıfırlama
    magnitudes = torch.abs(fft_complex)
    max_mag = torch.max(magnitudes)
    threshold = max_mag * threshold_ratio

    fft_filtered = fft_complex.clone()
    fft_filtered[magnitudes < threshold] = 0

    # Ters FFT (IDFT) ile zaman ortamına geri dönüşüm
    reconstructed = torch.fft.ifft(fft_filtered).real
    return reconstructed

# --- 4. 5. Görevdeki Sinyallerin Oluşturulması (N = 1024 = 2^10) ---
N = 1024
fs = 1000

# Sinyal 1: 50 Hz, Genlik 12
t_gpu, s1_clean = generate_signal(N, 12.0, 50.0, fs)
s1_noisy = add_noise(s1_clean, 0.35)

# Sinyal 2: 30 Hz, Genlik 5
_, s2_clean = generate_signal(N, 5.0, 30.0, fs)
s2_noisy = add_noise(s2_clean, 0.35)

# Toplam Sinyal
s_total_clean = s1_clean + s2_clean
s_total_noisy = s1_noisy + s2_noisy

# --- 5. Frekans Filtresinin Uygulanması ---
s1_filtered = frequency_threshold_filter(s1_noisy, threshold_ratio=0.20)
s2_filtered = frequency_threshold_filter(s2_noisy, threshold_ratio=0.20)
s_total_filtered = frequency_threshold_filter(s_total_noisy, threshold_ratio=0.20)

# --- 6. Görselleştirme ---
t = t_gpu.cpu().numpy()

fig, axes = plt.subplots(3, 1, figsize=(12, 9), sharex=True)

# 1. Sinyal
axes[0].plot(t[:200], s1_clean.cpu().numpy()[:200], 'b--', label='Orijinal Temiz (50 Hz)', alpha=0.7)
axes[0].plot(t[:200], s1_noisy.cpu().numpy()[:200], 'r', label='Gürültülü', alpha=0.4)
axes[0].plot(t[:200], s1_filtered.cpu().numpy()[:200], 'g', label='Frekans Filtreli (IDFT)', linewidth=2)
axes[0].set_title("1. Sinyal (f = 50 Hz, Genlik = 12)")
axes[0].grid(True)
axes[0].legend()

# 2. Sinyal
axes[1].plot(t[:200], s2_clean.cpu().numpy()[:200], 'b--', label='Orijinal Temiz (30 Hz)', alpha=0.7)
axes[1].plot(t[:200], s2_noisy.cpu().numpy()[:200], 'r', label='Gürültülü', alpha=0.4)
axes[1].plot(t[:200], s2_filtered.cpu().numpy()[:200], 'purple', label='Frekans Filtreli (IDFT)', linewidth=2)
axes[1].set_title("2. Sinyal (f = 30 Hz, Genlik = 5)")
axes[1].grid(True)
axes[1].legend()

# Toplam Sinyal
axes[2].plot(t[:200], s_total_clean.cpu().numpy()[:200], 'b--', label='Orijinal Temiz Toplam', alpha=0.7)
axes[2].plot(t[:200], s_total_noisy.cpu().numpy()[:200], 'r', label='Gürültülü Toplam', alpha=0.4)
axes[2].plot(t[:200], s_total_filtered.cpu().numpy()[:200], 'm', label='Frekans Filtreli Toplam (IDFT)', linewidth=2)
axes[2].set_title("Toplam Sinyal")
axes[2].set_xlabel("Zaman (saniye)")
axes[2].grid(True)
axes[2].legend()

plt.tight_layout()
plt.show()
```

```python
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt

# 1. GPU / CUDA Cihaz Ayarı
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# --- 7. Görev Fonksiyonu: 2D DFT ve Harmonik Spektrumu ---
def calculate_2d_dft_harmonics_gpu(image_path):
    """
    Girdi olarak resim yolunu alır, PyTorch CUDA ile 2D DFT hesaplar
    ve 2D harmonik spektrumunu (magnitüt) geri döndürür.
    """
    # Görseli gri tonlamalı (Grayscale) okuma
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise FileNotFoundError(f"{image_path} bulunamadı! Lütfen Colab dosyalar kısmına yüklediğinden emin ol.")

    # Görsel matrisini GPU'ya aktarma
    img_tensor = torch.tensor(img, dtype=torch.float32, device=device)

    # GPU üzerinde 2-Boyutlu FFT (2D DFT) hesaplama
    fft2_complex = torch.fft.fft2(img_tensor)

    # Düşük frekansları merkeze kaydırma (FFT Shift)
    fft2_shifted = torch.fft.fftshift(fft2_complex)

    # Harmonik Spektrumu (Magnitüt) hesaplama
    harmonic_spectrum_gpu = torch.abs(fft2_shifted)

    # Ekranda rahat görselleştirebilmek için Logaritmik Ölçekleme yapma
    log_harmonic_spectrum_gpu = torch.log(1 + harmonic_spectrum_gpu)

    # CPU / NumPy formatına dönüştürme
    return img, log_harmonic_spectrum_gpu.cpu().numpy()

# --- Uygulama / Test ---
image_file = 'resim3.jpeg'

# Fonksiyonu çalıştır
original_img, harmonic_spectrum = calculate_2d_dft_harmonics_gpu(image_file)

# --- Görselleştirme ---
plt.figure(figsize=(14, 6))

# Orijinal Görsel
plt.subplot(1, 2, 1)
plt.imshow(original_img, cmap='gray')
plt.title(f"Orijinal Görsel (Boyut: {original_img.shape[1]}x{original_img.shape[0]})")
plt.axis('off')

# 2D DFT Harmonik Spektrumu
plt.subplot(1, 2, 2)
plt.imshow(harmonic_spectrum, cmap='jet')
plt.title("2D DFT Harmonik Spektrumu (Frekans Ortamı)")
plt.colorbar(label='Logaritmik Genlik')
plt.axis('off')

plt.tight_layout()
plt.show()

print(f"CUDA Kullanıldı: {torch.cuda.is_available()}")
print(f"2D Harmonik Spektrum Matris Boyutu: {harmonic_spectrum.shape}")
```
