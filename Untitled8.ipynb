```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- 4. GÖREV: KENDİ BGR -> HSV DÖNÜŞÜM FONKSİYONUMUZ ---
def my_bgr2hsv(img_bgr):
    # Piksel değerlerini 0-1 aralığına getirip float32 tipine çeviriyoruz
    img_norm = img_bgr.astype(np.float32) / 255.0
    b = img_norm[:, :, 0]
    g = img_norm[:, :, 1]
    r = img_norm[:, :, 2]

    v = np.max(img_norm, axis=2)
    min_val = np.min(img_norm, axis=2)
    delta = v - min_val

    # Saturation (S) Hesaplama
    s = np.zeros_like(v)
    non_zero = v != 0
    s[non_zero] = delta[non_zero] / v[non_zero]

    # Hue (H) Hesaplama
    h = np.zeros_like(v)
    mask_r = (v == r) & (delta != 0)
    mask_g = (v == g) & (delta != 0)
    mask_b = (v == b) & (delta != 0)

    h[mask_r] = 60 * (((g[mask_r] - b[mask_r]) / delta[mask_r]) % 6)
    h[mask_g] = 60 * (((b[mask_g] - r[mask_g]) / delta[mask_g]) + 2)
    h[mask_b] = 60 * (((r[mask_b] - g[mask_b]) / delta[mask_b]) + 4)

    # OpenCV formatına uygun ölçekleme: H (0-179), S (0-255), V (0-255)
    h_cv = np.round(h / 2.0).astype(np.uint8)
    s_cv = np.round(s * 255.0).astype(np.uint8)
    v_cv = np.round(v * 255.0).astype(np.uint8)

    return cv2.merge([h_cv, s_cv, v_cv])


# --- RESİM OKUMA VE İŞLEME ---
img = cv2.imread('resim.jpg')

if img is None:
    print("HATA: 'resim.jpg' bulunamadı! Sol taraftaki klasöre 'resim.jpg' yüklediğinden emin ol.")
else:
    # 1. OpenCV hazır cvtColor fonksiyonu ile BGR -> HSV dönüşümü
    hsv_opencv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    # 2. Kendi fonksiyonumuz ile BGR -> HSV dönüşümü
    hsv_custom = my_bgr2hsv(img)

    # 3. Sayısal Kıyaslama ve Hata Değeri Hesaplama
    # İki görsel arasındaki mutlak fark alınır
    difference = np.abs(hsv_opencv.astype(np.int32) - hsv_custom.astype(np.int32))
    total_error = np.sum(difference)
    mean_error = np.mean(difference)

    print("=" * 55)
    print("--- 4. GÖREV RAPORU: RENK UZAYI DÖNÜŞÜMÜ (BGR -> HSV) ---")
    print(f"Toplam Mutlak Hata Değeri  : {total_error}")
    print(f"Piksel Başına Ortalama Hata : {mean_error:.4f}")
    if mean_error <= 1.0:
        print("SONUÇ: Başarılı! Hata değeri beklenildiği gibi 0'a çok yakın / 0.")
    print("=" * 55)

    # 4. Görselleri Ekranda Gösterme
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(hsv_opencv, cv2.COLOR_HSV2RGB))
    plt.title('OpenCV cvtColor Çıktısı (BGR -> HSV)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(hsv_custom, cv2.COLOR_HSV2RGB))
    plt.title('Kendi Fonksiyonumuzun Çıktısı')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Test İçin 5 Adet X-Ray Benzeri Görüntü Oluşturma
# (Eğer kendi 5 resmin varsa klasöre x1.jpg, x2.jpg... diye atıp cv2.imread ile okutabilirsin)
np.random.seed(42)
sample_images = []
for i in range(5):
    # X-Ray görüntüsü simülasyonu
    base = np.zeros((200, 200), dtype=np.uint8)
    cv2.circle(base, (100, 100), 60 + i*5, (150 + i*15), -1)
    cv2.rectangle(base, (30 + i*10, 40), (70 + i*10, 160), (200 - i*20), -1)
    noise = np.random.randint(0, 50, (200, 200), dtype=np.uint8)
    xray_img = cv2.add(base, noise)
    sample_images.append(xray_img)

# 2. İşleme ve Pseudocolor (Renk Paleti) Uygulama
# COLORMAP_JET veya COLORMAP_INFERNO güvenlik aramalarında en çok tercih edilen paletlerdir.
colored_images = []
for img in sample_images:
    # A) Zaten tek kanallı gri olan resmi belirginleştirme (Histogram Eşitleme)
    equalized = cv2.equalizeHist(img)
    # B) Güvenlik operatörleri için Renk Paleti (Pseudocolor) Uygulama
    colored = cv2.applyColorMap(equalized, cv2.COLORMAP_JET)
    colored_images.append(colored)

# 3. 5 Görüntüyü Yan Yana Görselleştirme
plt.figure(figsize=(15, 6))
for i in range(5):
    # Orijinal / Gri Görüntü
    plt.subplot(2, 5, i + 1)
    plt.imshow(sample_images[i], cmap='gray')
    plt.title(f'Görüntü {i+1} (Gri)')
    plt.axis('off')

    # Renklendirilmiş (Pseudocolor) Görüntü
    plt.subplot(2, 5, i + 6)
    plt.imshow(cv2.cvtColor(colored_images[i], cv2.COLOR_BGR2RGB))
    plt.title(f'Görüntü {i+1} (COLORMAP_JET)')
    plt.axis('off')

plt.tight_layout()
plt.show()

# 4. HOCAYA SUNULACAK AÇIKLAMA VE YORUM RAPORU
print("=" * 65)
print("--- 5. GÖREV: X-RAY GÖRÜNTÜ RENKLENDİRME VE RAPOR ---")
print("""
[RENK PALETİ SEÇİM NEDENİ]:
Güvenlik taramalarında 'COLORMAP_JET' (veya INFERNO/TURBO) paleti tercih edilmiştir.
İnsan gözü gri tonlarındaki çok küçük yoğunluk farklarını ayırt etmekte zorlanırken,
renk değişimlerine (mavi, yeşil, sarı, kırmızı) karşı çok daha duyarlıdır.

[YORUM VE FARK AÇIKLAMASI]:
1. Düşük Yoğunluklu / Yoğun Olmayan Maddeler (Mavi/Soğuk Tonlar): Valiz veya paketteki
   hafif nesneleri ve arka planı temsil eder.
2. Yüksek Yoğunluklu / Tehlikeli Maddeler (Sarı-Kırmızı/Sıcak Tonlar): Metal, sıvı
   veya kesici/delici nesneleri yoğunluk seviyesine göre parlatır.
3. Sonuç olarak sahte renklendirme (Pseudocolor), güvenlik operatörünün gizlenmiş
   nesneleri ve kontrast farklarını hızlıca fark etmesini sağlar.
""")
print("=" * 65)
```
