```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Hocanın 2. maddesi: cv2.calcHist kullanmadan kendi fonksiyonumuzu yazıyoruz
def my_calc_hist(channel):
    hist = np.zeros(256, dtype=int)
    for row in channel:
        for pixel in row:
            hist[pixel] += 1
    return hist

# Hocanın 1. maddesi: Renkli görüntüyü OpenCV ile okuma
img = cv2.imread('resim.jpg')

# Renkli resmi B, G, R kanallarına ayırma
b, g, r = cv2.split(img)

# Kendi fonksiyonumuzla her kanal için ayrı ayrı hesaplama
hist_b = my_calc_hist(b)
hist_g = my_calc_hist(g)
hist_r = my_calc_hist(r)

# Her kanalın histogramını ekran üzerinde çizdirme
colors = ('blue', 'green', 'red')
hists = (hist_b, hist_g, hist_r)
titles = ('Mavi (Blue) Kanalı Histogramı', 'Yeşil (Green) Kanalı Histogramı', 'Kırmızı (Red) Kanalı Histogramı')

for h, color, title in zip(hists, colors, titles):
    plt.figure(figsize=(6, 4))
    plt.title(title)
    plt.xlabel("Piksel Değeri (0-255)")
    plt.ylabel("Piksel Sayısı")
    plt.plot(h, color=color)
    plt.xlim([0, 256])
    plt.grid(True)
    plt.show()
```
