```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Kopyaladığınız resim yolunu buraya yazın
image_path = "resim.jpg"  # Örn: "/content/resim.jpg"

# Resmi okuma ve gri tonlamaya çevirme
img = cv2.imread(image_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Canny ile kenar tespiti
edges = cv2.Canny(gray, threshold1=50, threshold2=150)

# 3. Kenar resmine Median Filtre uygulanması
median_edges = cv2.medianBlur(edges, ksize=3)

# Çizim için kopyalar
img_hough_std = img.copy()
img_hough_prob = img.copy()

# 4.a) Standart Hough Dönüşümü (cv2.HoughLines)
lines_std = cv2.HoughLines(median_edges, rho=1, theta=np.pi/180, threshold=100)
if lines_std is not None:
    for line in lines_std:
        rho, theta = line[0]
        a, b = np.cos(theta), np.sin(theta)
        x0, y0 = a * rho, b * rho
        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * (a))
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * (a))
        cv2.line(img_hough_std, (x1, y1), (x2, y2), (0, 0, 255), 2)  # Kırmızı

# 4.b) Olasılıksal Hough Dönüşümü (cv2.HoughLinesP)
lines_prob = cv2.HoughLinesP(median_edges, rho=1, theta=np.pi/180, threshold=50, minLineLength=50, maxLineGap=10)
if lines_prob is not None:
    for line in lines_prob:
        x1, y1, x2, y2 = line[0]
        cv2.line(img_hough_prob, (x1, y1), (x2, y2), (0, 255, 0), 2)  # Yeşil

# 5. Görselleştirme
plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1), plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)), plt.title("Orijinal"), plt.axis("off")
plt.subplot(2, 3, 2), plt.imshow(edges, cmap="gray"), plt.title("Canny Kenarlar"), plt.axis("off")
plt.subplot(2, 3, 3), plt.imshow(median_edges, cmap="gray"), plt.title("Median Filtreli"), plt.axis("off")
plt.subplot(2, 3, 4), plt.imshow(cv2.cvtColor(img_hough_std, cv2.COLOR_BGR2RGB)), plt.title("HoughLines (Kırmızı)"), plt.axis("off")
plt.subplot(2, 3, 5), plt.imshow(cv2.cvtColor(img_hough_prob, cv2.COLOR_BGR2RGB)), plt.title("HoughLinesP (Yeşil)"), plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# GİRİŞ: Resim Okuma ve Ön İşleme
# ==========================================
# Colab panelinde sol tarafa yüklediğiniz resmin adı:
image_path = "resim.jpg"

img = cv2.imread(image_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# ==========================================
# MADDE 1: OpenCV Hazır Hough Dönüşümleri
# ==========================================
# 1. Canny ile kenar tespiti
edges = cv2.Canny(gray, threshold1=50, threshold2=150)

# 2. Median filtre uygulaması
median_edges = cv2.medianBlur(edges, ksize=3)

# Çizim kopyaları
img_hough_std = img.copy()
img_hough_prob = img.copy()

# 3.a) cv2.HoughLines (Standart - Kırmızı Çizgi)
lines_std = cv2.HoughLines(median_edges, rho=1, theta=np.pi/180, threshold=100)
if lines_std is not None:
    for line in lines_std:
        rho, theta = line[0]
        a, b = np.cos(theta), np.sin(theta)
        x0, y0 = a * rho, b * rho
        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * (a))
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * (a))
        cv2.line(img_hough_std, (x1, y1), (x2, y2), (0, 0, 255), 2)

# 3.b) cv2.HoughLinesP (Olasılıksal - Yeşil Çizgi)
lines_prob = cv2.HoughLinesP(median_edges, rho=1, theta=np.pi/180, threshold=50, minLineLength=50, maxLineGap=10)
if lines_prob is not None:
    for line in lines_prob:
        x1, y1, x2, y2 = line[0]
        cv2.line(img_hough_prob, (x1, y1), (x2, y2), (0, 255, 0), 2)

# ==========================================
# MADDE 2: Sıfırdan (Custom) Hough Line & Oylama Matrisi
# ==========================================
img_custom = img.copy()
height, width = median_edges.shape

# 1. Sin ve Cos değerlerini önceden hesaplayıp bellekte tutma (Lookup Table)
thetas = np.deg2rad(np.arange(-90, 90, 1))
cos_t = np.cos(thetas)
sin_t = np.sin(thetas)

max_rho = int(np.ceil(np.sqrt(height**2 + width**2)))
rhos = np.arange(-max_rho, max_rho, 1)

# 2. Oylama Matrisini (Accumulator Matrix) Oluşturma
accumulator = np.zeros((len(rhos), len(thetas)), dtype=np.int32)
y_idxs, x_idxs = np.nonzero(median_edges)

for x, y in zip(x_idxs, y_idxs):
    for t_idx in range(len(thetas)):
        rho = int(round(x * cos_t[t_idx] + y * sin_t[t_idx]))
        rho_idx = rho + max_rho
        accumulator[rho_idx, t_idx] += 1

# 3. Eşik Değeri (Threshold) ile Doğruları Bulma ve Çizme (Mavi Çizgi)
threshold = 120
rho_indices, theta_indices = np.where(accumulator >= threshold)

for r_idx, t_idx in zip(rho_indices, theta_indices):
    rho = rhos[r_idx]
    theta = thetas[t_idx]
    a, b = np.cos(theta), np.sin(theta)
    x0, y0 = a * rho, b * rho
    x1 = int(x0 + 1000 * (-b))
    y1 = int(y0 + 1000 * (a))
    x2 = int(x0 - 1000 * (-b))
    y2 = int(y0 - 1000 * (a))
    cv2.line(img_custom, (x1, y1), (x2, y2), (255, 0, 0), 2)

# Oylama matrisini resim olarak kaydetme (Ödev şartı)
acc_normalized = cv2.normalize(accumulator, None, 0, 255, cv2.NORM_MINMAX)
cv2.imwrite("oylama_matrisi.png", acc_normalized.astype(np.uint8))

# ==========================================
# GÖRSELLEŞTİRME: Tüm Sonuçları Yan Yana Göster
# ==========================================
plt.figure(figsize=(18, 10))

plt.subplot(2, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Orijinal Resim")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(median_edges, cmap="gray")
plt.title("Median Filtreli Kenarlar")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(accumulator, cmap='jet', aspect='auto')
plt.title("Oylama Matrisi (Accumulator)")
plt.colorbar()

plt.subplot(2, 3, 4)
plt.imshow(cv2.cvtColor(img_hough_std, cv2.COLOR_BGR2RGB))
plt.title("1. Madde: cv2.HoughLines (Kırmızı)")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(cv2.cvtColor(img_hough_prob, cv2.COLOR_BGR2RGB))
plt.title("1. Madde: cv2.HoughLinesP (Yeşil)")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(cv2.cvtColor(img_custom, cv2.COLOR_BGR2RGB))
plt.title("2. Madde: Sıfırdan Hough Line (Mavi)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Resim ve Kenar Görüntüsünü Hazırlama
img_custom_circle = img.copy()
edges_input = median_edges  # Önceden ürettiğimiz median filtreli kenarlar

height, width = edges_input.shape

# 2. Sinus ve Cosinus Lookup Table (Açı: 0 - 360 derece)
angles = np.arange(0, 360, 5)  # Performans için 5'er derecelik adımlarla
rads = np.deg2rad(angles)
cos_a = np.cos(rads)
sin_a = np.sin(rads)

# Aranacak yarıçap (r) aralığı
r_min = 20
r_max = 80
r_step = 2
radii = np.arange(r_min, r_max, r_step)

# 3. 3-Boyutlu Oylama Matrisi (Accumulator 3D: [Height, Width, Radii])
accumulator_3d = np.zeros((height, width, len(radii)), dtype=np.int32)

# Kenar olan piksellerin (255) koordinatlarını alma
y_idxs, x_idxs = np.nonzero(edges_input)

# Oylama İşlemi (Hough Circle Transform Oylaması)
for x, y in zip(x_idxs, y_idxs):
    for r_idx, r in enumerate(radii):
        for c, s in zip(cos_a, sin_a):
            # Çember denklemi: a = x - r*cos(theta), b = y - r*sin(theta)
            a = int(round(x - r * c))
            b = int(round(y - r * s))

            # Koordinatların resim sınırları içinde olup olmadığını kontrol etme
            if 0 <= a < width and 0 <= b < height:
                accumulator_3d[b, a, r_idx] += 1  # (y, x, r_idx)

# 4. Eşik Değeri (Threshold) İle Çemberleri Tespit Etme
threshold = 30  # Çember yoğunluğuna göre değiştirebilirsiniz
y_indices, x_indices, r_indices = np.where(accumulator_3d >= threshold)

# Üst üste binen çok fazla çember çizimini engellemek için basit basıtlama (non-max suppression benzeri)
detected_circles = []
for y, x, r_idx in zip(y_indices, x_indices, r_indices):
    r = radii[r_idx]
    # Yakın merkezli çemberleri eleme
    too_close = False
    for cx, cy, cr in detected_circles:
        if np.sqrt((x - cx)**2 + (y - cy)**2) < 15:
            too_close = True
            break
    if not too_close:
        detected_circles.append((x, y, r))

# Çemberleri resim üzerine çizme (Mavi Çember, Kırmızı Merkez)
for x, y, r in detected_circles:
    cv2.circle(img_custom_circle, (x, y), r, (255, 0, 0), 2)
    cv2.circle(img_custom_circle, (x, y), 2, (0, 0, 255), 3)

# 5. Görselleştirme ve Kaydetme
plt.figure(figsize=(14, 6))

# 3D Oylama Matrisinin Yarıçap Boyunca En Yüksek Değerlerinin İzdüşümü (2D Gösterimi)
plt.subplot(1, 2, 1)
acc_max_projection = np.max(accumulator_3d, axis=2)
plt.imshow(acc_max_projection, cmap='hot')
plt.title("3D Oylama Matrisi İzdüşümü (Hough Space)")
plt.colorbar()

# Tespit Edilen Çemberlerin Olduğu Resim
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_custom_circle, cv2.COLOR_BGR2RGB))
plt.title("4. Madde: Sıfırdan Hough Circle Sonucu")
plt.axis("off")

plt.tight_layout()
plt.show()

# Oylama matrisi izdüşümünü resim olarak kaydetme (Ödev şartı)
acc_norm = cv2.normalize(acc_max_projection, None, 0, 255, cv2.NORM_MINMAX)
cv2.imwrite("4_oylama_matrisi_3d.png", acc_norm.astype(np.uint8))
cv2.imwrite("4_custom_hough_circles.png", img_custom_circle)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Resim ve Kenar Görüntüsünü Hazırlama
img_custom_circle = img.copy()
edges_input = median_edges

height, width = edges_input.shape

# 2. Sinus ve Cosinus değerlerini vektörel hesaplama (Lookup Table)
angles = np.arange(0, 360, 10)  # 10'ar derecelik adımlarla hızlandırma
rads = np.deg2rad(angles)
cos_a = np.cos(rads)
sin_a = np.sin(rads)

# Yarıçap (r) aralığı
r_min, r_max, r_step = 20, 80, 2
radii = np.arange(r_min, r_max, r_step)

# 3. 3D Oylama Matrisi (Accumulator 3D)
accumulator_3d = np.zeros((height, width, len(radii)), dtype=np.int32)

y_idxs, x_idxs = np.nonzero(edges_input)

# Vektörel Hızlı Oylama İşlemi (İç döngüler kaldırıldı)
for r_idx, r in enumerate(radii):
    # Tüm acilar ve tum kenar pikselleri için (a, b) merkez matrislerini tek seferde üretme
    a_matrix = (x_idxs[:, None] - r * cos_a[None, :]).round().astype(int)
    b_matrix = (y_idxs[:, None] - r * sin_a[None, :]).round().astype(int)

    # Sadece resim sınırları içinde kalan noktaları alma
    valid_mask = (a_matrix >= 0) & (a_matrix < width) & (b_matrix >= 0) & (b_matrix < height)

    valid_a = a_matrix[valid_mask]
    valid_b = b_matrix[valid_mask]

    # Matris indeksine hızlı ekleme (np.add.at)
    np.add.at(accumulator_3d[:, :, r_idx], (valid_b, valid_a), 1)

# 4. Eşik Değeri (Threshold) İle Çemberleri Tespit Etme
threshold = 30
y_indices, x_indices, r_indices = np.where(accumulator_3d >= threshold)

# Yakın çemberleri eleme
detected_circles = []
for y, x, r_idx in zip(y_indices, x_indices, r_indices):
    r = radii[r_idx]
    too_close = False
    for cx, cy, cr in detected_circles:
        if np.sqrt((x - cx)**2 + (y - cy)**2) < 15:
            too_close = True
            break
    if not too_close:
        detected_circles.append((x, y, r))

# Çemberleri resim üzerine çizme
for x, y, r in detected_circles:
    cv2.circle(img_custom_circle, (x, y), r, (255, 0, 0), 2)  # Mavi Çember
    cv2.circle(img_custom_circle, (x, y), 2, (0, 0, 255), 3)  # Kırmızı Merkez

# 5. Görselleştirme ve Kaydetme
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
acc_max_projection = np.max(accumulator_3d, axis=2)
plt.imshow(acc_max_projection, cmap='hot')
plt.title("3D Oylama Matrisi İzdüşümü (Hough Space)")
plt.colorbar()

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_custom_circle, cv2.COLOR_BGR2RGB))
plt.title("4. Madde: Hızlı Sıfırdan Hough Circle Sonucu")
plt.axis("off")

plt.tight_layout()
plt.show()

# Kaydetme
acc_norm = cv2.normalize(acc_max_projection, None, 0, 255, cv2.NORM_MINMAX)
cv2.imwrite("4_oylama_matrisi_3d.png", acc_norm.astype(np.uint8))
cv2.imwrite("4_custom_hough_circles.png", img_custom_circle)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Resim ve Kenar Görüntüsü
img_concentric = img.copy()
edges_input = median_edges  # 1. adımdan gelen median filtreli kenarlar

height, width = edges_input.shape

# 2. Sinus ve Cosinus Lookup Table
angles = np.arange(0, 360, 10)  # Hızlı işlem için 10 derecelik adımlar
rads = np.deg2rad(angles)
cos_a = np.cos(rads)
sin_a = np.sin(rads)

# Yarıçap (r) aralığı
r_min, r_max, r_step = 15, 90, 3
radii = np.arange(r_min, r_max, r_step)
num_radii = len(radii)

# 3. 3-Boyutlu Oylama Matrisi (Accumulator 3D)
accumulator_3d = np.zeros((height, width, num_radii), dtype=np.int32)
y_idxs, x_idxs = np.nonzero(edges_input)

# Vektörel Hızlı Oylama
for r_idx, r in enumerate(radii):
    a_matrix = (x_idxs[:, None] - r * cos_a[None, :]).round().astype(int)
    b_matrix = (y_idxs[:, None] - r * sin_a[None, :]).round().astype(int)

    valid_mask = (a_matrix >= 0) & (a_matrix < width) & (b_matrix >= 0) & (b_matrix < height)

    valid_a = a_matrix[valid_mask]
    valid_b = b_matrix[valid_mask]

    np.add.at(accumulator_3d[:, :, r_idx], (valid_b, valid_a), 1)

# 4. EŞ MERKEZLİ ÇEMBER TESPİTİ (Oylama Matrisi Üzerinden Doğrudan Mantık)
# Bir (y, x) merkezinin en az 'min_radius_count' sayıda farklı yarıçapta eşik değerini (single_r_thresh) geçmesi gerekir.
single_r_thresh = 20    # Tek bir yarıçap için kenar oyu eşiği
min_radius_count = 2    # Eş merkezli kabul edilmek için en az kaç farklı yarıçapta çember olmalı

# (height, width, num_radii) matrisinde eşiği geçen hücreler (1: Var, 0: Yok)
votes_binary = (accumulator_3d >= single_r_thresh).astype(int)

# Her bir merkez (y, x) için kaç farklı yarıçapta çember olduğunu hesaplama
concentric_center_counts = np.sum(votes_binary, axis=2)

# En az 2 veya daha fazla eş merkezli yarıçapa sahip merkez koordinatları
concentric_y_idxs, concentric_x_idxs = np.where(concentric_center_counts >= min_radius_count)

# Tespit edilen eş merkezli çemberleri çizme
detected_concentric_circles = []

for y, x in zip(concentric_y_idxs, concentric_x_idxs):
    # Bu merkezde tespit edilen yarıçapları bulma
    valid_r_indices = np.where(accumulator_3d[y, x, :] >= single_r_thresh)[0]

    # Çok yakın merkez tekrarlarını önlemek için kontrol
    too_close = False
    for cx, cy, _ in detected_concentric_circles:
        if np.sqrt((x - cx)**2 + (y - cy)**2) < 12:
            too_close = True
            break

    if not too_close:
        for r_idx in valid_r_indices:
            r = radii[r_idx]
            detected_concentric_circles.append((x, y, r))

# Çizim (Sarı Eş Merkezli Çemberler, Kırmızı Merkez)
for x, y, r in detected_concentric_circles:
    cv2.circle(img_concentric, (x, y), r, (0, 255, 255), 2)  # Sarı renkli çemberler
    cv2.circle(img_concentric, (x, y), 2, (0, 0, 255), 3)    # Kırmızı merkez noktası

# 5. Görselleştirme ve Kaydetme
plt.figure(figsize=(14, 6))

# Eş Merkezli Çember Oylama İzdüşümü
plt.subplot(1, 2, 1)
plt.imshow(concentric_center_counts, cmap='magma')
plt.title("Eş Merkezli Oylama Haritası (Concentric Map)")
plt.colorbar()

# Tespit Edilen Eş Merkezli Çemberler
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_concentric, cv2.COLOR_BGR2RGB))
plt.title("5. Madde: Eş Merkezli Çember Sonucu")
plt.axis("off")

plt.tight_layout()
plt.show()

# Oylama görselini resim olarak kaydetme (Ödev şartı)
acc_norm = cv2.normalize(concentric_center_counts, None, 0, 255, cv2.NORM_MINMAX)
cv2.imwrite("5_concentric_accumulator.png", acc_norm.astype(np.uint8))
cv2.imwrite("5_concentric_circles_result.png", img_concentric)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import math
from numba import cuda

# ==========================================
# 1. CUDA KERNEL (Paralel Hough Line Fonksiyonu)
# ==========================================
@cuda.jit
def hough_line_kernel(d_edges, d_cos_t, d_sin_t, d_accumulator, height, width, num_thetas, max_rho):
    # GPU üzerindeki her bir izlek (thread) için 2D indeks hesabı
    x, y = cuda.grid(2)

    # Resim sınırları içindeysek ve piksel kenar pikseliyse (255 ise)
    if x < width and y < height:
        if d_edges[y, x] > 0:
            # Her bir açı (theta) için paralel/döngüsel hesaplama
            for t_idx in range(num_thetas):
                # rho = x * cos(theta) + y * sin(theta)
                rho = int(round(x * d_cos_t[t_idx] + y * d_sin_t[t_idx]))
                rho_idx = rho + max_rho

                # Atomik Ekleme (Atomic Add): Birden fazla thread aynı matris hücresine
                # aynı anda yazmaya çalıştığında veri kaybını engeller.
                cuda.atomic.add(d_accumulator, (rho_idx, t_idx), 1)

# ==========================================
# 2. ANA KOD VE VERİ TRANSFERİ (HOST TO DEVICE)
# ==========================================
img_cuda = img.copy()
edges_input = median_edges  # Önceki adımlardan gelen kenar görüntüsü
height, width = edges_input.shape

# Theta ve Rho Parametreleri
thetas = np.deg2rad(np.arange(-90, 90, 1), dtype=np.float32)
num_thetas = len(thetas)

cos_t = np.cos(thetas).astype(np.float32)
sin_t = np.sin(thetas).astype(np.float32)

max_rho = int(np.ceil(np.sqrt(height**2 + width**2)))
rhos = np.arange(-max_rho, max_rho, 1)
num_rhos = len(rhos)

# Host (CPU) Belleğinde Matrisler
accumulator_cpu = np.zeros((num_rhos, num_thetas), dtype=np.int32)

# --- Verileri GPU (Device) Belleğine Aktarma ---
d_edges = cuda.to_device(edges_input)
d_cos_t = cuda.to_device(cos_t)
d_sin_t = cuda.to_device(sin_t)
d_accumulator = cuda.to_device(accumulator_cpu)

# CUDA Blok ve Izgara (Grid) Yapısını Tanımlama
threads_per_block = (16, 16)
blocks_per_grid_x = math.ceil(width / threads_per_block[0])
blocks_per_grid_y = math.ceil(height / threads_per_block[1])
blocks_per_grid = (blocks_per_grid_x, blocks_per_grid_y)

# ==========================================
# 3. KERNEL'İ GPU ÜZERİNDE ÇALIŞTIRMA
# ==========================================
hough_line_kernel[blocks_per_grid, threads_per_block](
    d_edges, d_cos_t, d_sin_t, d_accumulator, height, width, num_thetas, max_rho
)

# Sonucu GPU'dan CPU'ya Geri Kopyalama (Device to Host)
accumulator_cuda = d_accumulator.copy_to_host()

# ==========================================
# 4. EŞİK DEĞERİ İLE DOĞRULARI BAZ ALIP ÇİZME
# ==========================================
threshold = 120
rho_indices, theta_indices = np.where(accumulator_cuda >= threshold)

for r_idx, t_idx in zip(rho_indices, theta_indices):
    rho = rhos[r_idx]
    theta = thetas[t_idx]
    a, b = np.cos(theta), np.sin(theta)
    x0, y0 = a * rho, b * rho
    x1 = int(x0 + 1000 * (-b))
    y1 = int(y0 + 1000 * (a))
    x2 = int(x0 - 1000 * (-b))
    y2 = int(y0 - 1000 * (a))
    cv2.line(img_cuda, (x1, y1), (x2, y2), (255, 0, 255), 2)  # Mor renk çizgi

# ==========================================
# 5. GÖRSELLEŞTİRME VE KAYDETME
# ==========================================
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.imshow(accumulator_cuda, cmap='jet', aspect='auto')
plt.title("CUDA Oylama Matrisi (Accumulator)")
plt.xlabel("Theta Index")
plt.ylabel("Rho Index")
plt.colorbar()

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_cuda, cv2.COLOR_BGR2RGB))
plt.title("CUDA Paralel Hough Line Sonucu")
plt.axis("off")

plt.tight_layout()
plt.show()

# Resimleri kaydetme
acc_norm = cv2.normalize(accumulator_cuda, None, 0, 255, cv2.NORM_MINMAX)
cv2.imwrite("cuda_oylama_matrisi.png", acc_norm.astype(np.uint8))
cv2.imwrite("cuda_hough_line_result.png", img_cuda)
```
